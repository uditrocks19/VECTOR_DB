```python
# lets understand name entity recognition (NER) and how to use it with LLMs. NER is a subtask of information extraction that seeks to locate and classify named entities mentioned in unstructured text into pre-defined categories such as the names of persons, organizations, locations, expressions of times, quantities, monetary values, percentages, etc.
```

```python
load_dotenv()  # take environment variables from .env.
```

```python
from transformers import AutoTokenizer, AutoModelForTokenClassification
import torch
from transformers import pipeline
```

```python
model_id = "dslim/bert-base-NER"
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_id) # huggingface model hub
```

```python
ner_model = AutoModelForTokenClassification.from_pretrained(model_id)
```

```python
device = 'cpu'
```

```python
# nlp pipeline for NER
nlp = pipeline("ner", model=ner_model, tokenizer=tokenizer, device=device, aggregation_strategy="max")
```

```python
nlp("Udit is my name and I live in India. I work at Microsoft and my email is udit@example.com")
```

```python
nlp("I love AI")
```
